```python
import pandas as pd
import matplotlib
matplotlib.use("Agg")  #safe for headless/script evacuation
import matplotlib.pyplot as plt
import os

DATA_PATH = "Investment survey dataset.csv"
CHART_DIR = "Charts"
os.makedirs(CHART_DIR, exist_ok=True)

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)
```

```python
#Step 1: Load the dataset
df = pd.read_csv(DATA_PATH)

#  columns names
df.columns = (
    df.columns.str.strip()
    .str.replace(" ", "_")
    .str.replace("?","", regex=False)
)
# The savings-objections columns becomes: What_are_your_savings_objectives

#Fix the misspelled columns so code is readable
df = df.rename(columns={"Stock_Market":"Stock_Market"})
obj_cols = df.select_dtypes(include="object").columns
for c in obj_cols:
  df[c] = df[c].str.strip()

#Check for the missing values and duplicates
print("=== Missing values per column ===")
print(df.isnull().sum()[df.isnull().sum() > 0])
print("\nDuplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()

#Sanity check data types
print("=== Data types ===")
print(df.dtypes)

print("\nCleaned shape:", df.shape)
print("="*60)


```

```python
df.info()
df.head()
list(df.columns)
```

```python
#Task 2: Gender Distribution
gender_counts = df["gender"].value_counts()
print(gender_counts)

plt.figure(figsize=(5, 5))
gender_counts.plot(kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("Gender Distribution of Respondents")
plt.ylabel("")
plt.tight_layout()
plt.savefig(f"{CHART_DIR}/task2_gender_distribution.png")
plt.close()
print(f"Saved chart -> {CHART_DIR}/task2_gender_distribution.png")
```

```python
#Task 3:Discriptive Statistics
numeric_cols = df.select_dtypes(include="number").columns
print("Numeric columns detected:", list(numeric_cols))
desc_stats = df[numeric_cols].agg(["mean", "median", "std"]).T
print(desc_stats)
desc_stats.to_csv(f"{CHART_DIR}/../task3_descriptive_stats.csv")
```

```python
#Task 4: Most Preferred Investment Avenue
avenue_counts= df["Avenue"].value_counts()
print(avenue_counts)
print("\nMost preferred avenue:", avenue_counts.idxmax(),
      f"({avenue_counts.max()} respondents)")

plt.figure(figsize=(7,4))
avenue_counts.plot(kind="bar", color = "#E29A5E")
plt.title("Most Preferred Investment Avenue")
plt.xlabel("Avenue")
plt.ylabel("Number of Respondents")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.savefig(f"{CHART_DIR}/task4_most_preferred_avenue.png")
plt.close()
print(f"Saved chart -> {CHART_DIR}/task4_most_preferred_avenue.png")
```

```python
#Task 5: Reason for Investments
reason_cols = ["Reason_Equity", "Reason_Mutual", "Reason_Bonds", "Reason_FD"]
for cols in reason_cols:
  print(f"\n-- {cols} --")
  print(df[cols].value_counts())
```

```python
#Task 6: Savings Objective
obj_cols = "What_are_your_savings_objectives"
objective_counts = df[obj_cols].value_counts()
print(objective_counts)

plt.figure(figsize=(7, 4))
objective_counts.plot(kind="bar", color="#7FB88E")
plt.title("Main Savings Objective")
plt.xlabel("Objective")
plt.ylabel("Count")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.savefig(f"{CHART_DIR}/task6_savings_objective.png")
plt.close()
print(f"Saved chart -> {CHART_DIR}/task6_savings_objective.png")
```

```python
#Task 7: Common Information Source
source_counts = df["Source"].value_counts()
print(source_counts)

plt.figure(figsize=(7, 4))
source_counts.plot(kind="bar", color="#C9A6E8")
plt.title("Where Investors Get Their Information")
plt.xlabel("Source")
plt.ylabel("Count")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.savefig(f"{CHART_DIR}/task7_info_source.png")
plt.close()
print(f"Saved chart -> {CHART_DIR}/task7_info_source.png")
```

```python
#Task 8: Investment Duration
duration_counts = df["Duration"].value_counts()
print(duration_counts)

duration_map = {
    "Less than 1 year": 0.5,
    "1-3 years": 2,
    "3-5 years": 4,
    "More than 5 years": 6,
}
df["Duration_numeric"] = df["Duration"].map(duration_map)
avg_duration = df["Duration_numeric"].mean()
print("f\nMost common duration bucket: {duration_counts.idxmax()}")
print(f"Approx. average investment duration (year, midpoint method): {avg_duration:.2f}")
```

```python
#Task 9: Expectations from Investments
expect_counts = df["Expect"].value_counts()
print(expect_counts)

plt.figure(figsize=(6, 4))
expect_counts.plot(kind ="bar", color="#5EA3E2")
plt.title("Expected Returns  from Investments")
plt.xlabel("Expected Returns Range")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig(f"{CHART_DIR}/task9_expections.png")
plt.close()
print(f"Saved chart -> {CHART_DIR}/task9_expections.png")
```

```python
#Task 10: Correlation Analysis
expect_map = {"10%-20%": 15, "20%-30%": 25, "30%-40%": 35}
df["Expect_numeric"] = df["Expect"].map(expect_map)

corr_df = df[["age", "Duration_numeric", "Expect_numeric",
              "Mutual_Funds", "Equity_Market", "Fixed_Deposits", "Gold"]]
correlation_matrix = corr_df.corr()
print(correlation_matrix.round(2))

plt.figure(figsize=(7, 6))
plt.imshow(correlation_matrix, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Correlation")
plt.xticks(range(len(correlation_matrix.columns)), correlation_matrix.columns, rotation=45, ha="right")
plt.yticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)
for i in range(len(correlation_matrix.columns)):
  for j in range(len(correlation_matrix.columns)):
    plt.text(j, i, f"{correlation_matrix.iloc[i, j]:.2f}", ha="center", va="center", fontsize=8)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig(f"{CHART_DIR}/task10_correlation_heatmap.png")
plt.close()
print(f"Saved chart -> {CHART_DIR}/task10_correlation_heatmap.png")

print("\nAll tasks complete.Check the 'charts folder for visuals.")
```
